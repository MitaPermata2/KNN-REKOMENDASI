```python
import pandas as pd
import random

# Fungsi membuat kombinasi
def generate_combinations(df, preferensi=None):
    kategori_list = ['makan_pagi', 'makan_siang', 'makan_malam', 'snack_1', 'snack_2']
    kombinasi = []
    total_natrium = 0.0

    for kategori in kategori_list:
        df_kategori = df[df['kategori'] == kategori]

        # Cek preferensi makanan
        if preferensi:
            df_pref = df_kategori[df_kategori['nama_makanan'].isin(preferensi)]
            if not df_pref.empty:
                selected = df_pref.sample(1)
            else:
                selected = df_kategori.sample(1)
        else:
            selected = df_kategori.sample(1)
        
        # Pastikan 'selected' bukan kosong dan natrium terkonversi dengan benar
        if not selected.empty:
            natrium_value = pd.to_numeric(selected['natrium (mg)'].values[0], errors='coerce')
            if pd.notna(natrium_value):
                total_natrium += natrium_value
            
            kombinasi.append(selected)

    hasil = pd.concat(kombinasi).reset_index(drop=True)
    return hasil, total_natrium

# Buat kombinasi makanan otomatis
samples = []
total_natriums = []

for _ in range(1000):
    combo, total_natrium = generate_combinations(df)
    samples.append(combo)  # Simpan hasil kombinasi makanan
    total_natriums.append(total_natrium)  # Simpan total natrium

# Memeriksa hasil kombinasi makanan dan total natrium
print(samples[0])  # Tampilkan kombinasi makanan pertama
print(total_natriums[:5])  # Tampilkan total natrium untuk 5 iterasi pertama
```

```python
#simpan hasil kombinasi ke dalam DataFrame
kombinasi_data = []

for idx, (kombo, total) in enumerate(zip(samples, total_natriums)):
    row = {
        'Id_kombinasi': idx+1,
        'Makan_pagi': kombo[kombo['kategori'] == 'makan_pagi']['nama_makanan'].values[0],
        'Makan_siang': kombo[kombo['kategori'] == 'makan_siang']['nama_makanan'].values[0],
        'Makan_malam': kombo[kombo['kategori'] == 'makan_malam']['nama_makanan'].values[0],
        'Snack_1': kombo[kombo['kategori'] == 'snack_1']['nama_makanan'].values[0],
        'Snack_2': kombo[kombo['kategori'] == 'snack_2']['nama_makanan'].values[0],
        'Total_natrium': total
    }
    kombinasi_data.append(row)

df_kombinasi = pd.DataFrame(kombinasi_data)
```

```python
#simpan ke csv
df_kombinasi.to_csv('kombinasi_menu.csv', index=False)
print("Hasil kombinasi makanan telah disimpan ke dalam 'kombinasi_makanan.csv'")
```
